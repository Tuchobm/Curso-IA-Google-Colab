## 1. Indentación en Python

La indentación es crucial en Python, ya que define la estructura y el bloque de código que pertenece a cada declaración. Aunque el número de espacios puede variar, se recomienda usar **4 espacios** por nivel. Una indentación inconsistente puede provocar errores de sintaxis.

````python
# Ejemplo de estructura condicional con indentación correcta
numero = 5

if numero > 0:
    print("El número es positivo")
else:
    print("El número es negativo o cero")
````

## 2. Variables en Python
Las variables son contenedores que almacenan valores. En programación, una variable es un nombre asignado a un valor. Este valor puede ser de diferentes tipos (número entero, número decimal, texto, lista, etc.), y en el caso de Python, no necesitas declarar explícitamente el tipo de dato antes de usar la variable.

Si la variable es de tipo texto (también llamado cadena o string) el valor irá entre comillas dobles o simples.

**Ejemplo práctico:**

````python
# Declaración de variables
mi_variable = 10
print("El valor de mi_variable es:", mi_variable)

mi_variable_cadena = "texto de prueba"
print("El valor de mi_variable es:", mi_variable_cadena)
````

**Tipos de datos básicos en Python:**
- Enteros (int)
- Flotantes (float)
- Cadenas (str)
- Booleanos (bool)

## 3. Listas
Una lista en Python es una colección ordenada de elementos. Puede contener valores de diferentes tipos, como números, texto y objetos. Se utiliza para almacenar y manipular datos de manera flexible y dinámica. Se definen utilizando corchetes [].

````python
# Ejemplo de listas
mi_lista = [1, 2, 3, 4]
print("Mi lista:", mi_lista)
print("El primer elemento es:", mi_lista[0])
print("La longitud de la lista es:", len(mi_lista))
````

**Operaciones con listas:**
- Concatenación: +
- Repetición: *
- Modificación de elementos: [] = 
- Métodos como `append()`, `remove()`, `sort()`

## 4. Sets
Un set en Python es una colección desordenada de elementos únicos. No permite elementos duplicados y no garantiza un orden específico de los elementos. Se definen utilizando llaves `{ }`.

````python
# Ejemplo de set
mi_set = {1, 2, 3, 3, 'a', 'b'}
print("Mi set:", mi_set)
````

**Operaciones con sets:**
- Agregar elementos: `add()`
- Eliminar elementos: `remove()`
- Comprobar pertenencia: `in`

## 5. Tuplas
Una tupla en Python es una colección ordenada e inmutable de elementos. Una vez definida, no se puede modificar. Se definen utilizando paréntesis `( )`.

````python
# Ejemplo de tupla
mi_tupla = (1, 2, 3, 'a', 'b')
print("Mi tupla:", mi_tupla)
````

**Operaciones con tuplas:**
- Acceso a elementos: `[]`
- Concatenación: +
- Repetición: *

## 6. Condicionales
Las estructuras condicionales permiten evaluar condiciones y tomar decisiones para dirigir el flujo del programa. Esto permite la creación de lógica compleja y personalizada en función de datos variables. De esta manera, se puede controlar qué acciones se realizan o no según las circunstancias específicas del programa.

**Sintaxis básica:**
```python
if condición:
    # Código que se ejecuta si la condición es verdadera
else:
    # Código que se ejecuta si la condición es falsa
```


````python
# Ejemplo de condicionales
num = 7
if num > 5:
    print("El número es mayor a 5")
else:
    print("El número es menor o igual a 5")
````

## 7. Bucles

Los bucles permiten ejecutar un bloque de código de forma repetitiva. En Python existen principalmente dos tipos:

- Bucle `while`: ejecuta el código mientras una condición sea verdadera.
- Bucle `for`: itera sobre elementos de una secuencia (listas, cadenas, etc.).

También se pueden anidar bucles para recorrer estructuras de datos multidimensionales.

````python
# Ejemplo: sumar números del 1 al 10 usando un bucle while
n = 10
suma = 0
contador = 1

while contador <= n:
    suma = suma + contador
    contador = contador + 1

print("La suma de 1 a", n, "es", suma)
````

````python
# Ejemplo: iterar sobre una lista de animales utilizando un bucle for
animales = ["perro", "gato", "loro"]

for animal in animales:
    print("Este es un", animal)
````

````python
# Ejemplo: bucle anidado que imprime pares de números
for i in range(1, 4):
    for j in range(1, 4):
        print(f"i: {i}, j: {j}")
````

## 8. Declaración de funciones

Las funciones en Python son bloques de código reutilizables que se definen con la palabra clave `def`. Permiten agrupar instrucciones, recibir parámetros y devolver resultados usando la palabra clave `return`. Además, Python soporta funciones anónimas (lambda) para operaciones simples.

````python
# Ejemplo: función que saluda a un usuario
def saludar(nombre):
    mensaje = f"¡Hola, {nombre}!"
    return mensaje

# Llamamos a la función y mostramos el resultado
resultado = saludar("Carlos")
print(resultado)
````

````python
# Ejemplo: uso de una función lambda para sumar dos números
suma = lambda a, b: a + b
print("La suma es:", suma(3, 5))
````

## 9. Importación de librerías

Python permite extender sus funcionalidades mediante librerías. Puedes importar módulos estándar, bibliotecas externas instaladas con `pip`, o incluso módulos personalizados creados por ti.

Existen distintas formas de importar módulos en Python:

- `import nombre_modulo`: importa todo el módulo.
- `from nombre_modulo import funcion`: importa una función o clase específica.
- `import nombre_modulo as alias`: importa un módulo con un alias.
- `from nombre_modulo import *`: importa todas las funciones y clases de un módulo (no recomendado en muchos casos)

````python
# Ejemplo: importar el módulo math y usar sus funciones
import math

print("Raíz cuadrada de 16:", math.sqrt(16))
print("Valor de pi:", math.pi)
````

````python
# Ejemplo: importar solo la función sqrt de math
from math import sqrt

print("Raíz cuadrada de 25:", sqrt(25))
````

````python
# Ejemplo: importar numpy con un alias
import numpy as np

array = np.array([1, 2, 3, 4])
print("Array de numpy:", array)
````
