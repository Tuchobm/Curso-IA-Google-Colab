```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

## Cargar datos

Primero deberéis descargar los datos de la competición y descomprimirlos en una carpeta local. Una vez descargados podéis subirlos a Google Colab o a cualquier otro entorno de trabajo que utilicéis.

A continuación se cargan los datos de entrenamiento y de test. Es fundamental comprender la estructura y las posibles transformaciones que se deben aplicar antes de entrenar el modelo.

```python
train = pd.read_csv("train.csv")
test = pd.read_csv("test.csv")
```

## Transformar datos

```python
from sklearn.model_selection import train_test_split

train, val = train_test_split(train, test_size=0.2, random_state=0)
train = train.reset_index(drop=True)
```

## Entrenar modelo

```python
from sklearn.metrics import accuracy_score


def transform_data(df):
    """Transforma los datos para el modelo."""
    # Eliminar columnas no necesarias
    df = df.drop(columns=["id", "url"])
    if "label" in df.columns:
        df = df.drop(columns=["label"])

    # Eliminar datos no numéricos
    df = df.select_dtypes(include=[np.number])

    # Manejar valores faltantes
    df = df.fillna(df.mean())

    return df


def train_model(model):
    """Entrena el modelo con los datos de entrenamiento y evalúa el rendimiento."""
    # Transformar los datos de entrenamiento y validación
    X_train = transform_data(train)
    y_train = train["label"]

    X_val = transform_data(val)
    y_val = val["label"]

    # Entrenar el modelo
    model.fit(X_train, y_train)
    print("Modelo entrenado.")

    # Calcular las predicciones y evaluar el entrenamiento
    y_pred_train = model.predict(X_train)
    acc_train = accuracy_score(y_train, y_pred_train)
    print(f"Accuracy entrenamiento: {acc_train:.2f}")

    # Calcular y evaluar el conjunto de validación
    y_pred_val = model.predict(X_val)
    acc_val = accuracy_score(y_val, y_pred_val)
    print(f"Accuracy validación: {acc_val:.2f}")

    # Consejo para los alumnos: experimente con diferentes modelos o ajuste de hiperparámetros.
```

```python
from sklearn.linear_model import LogisticRegression

model = LogisticRegression(random_state=0)
train_model(model)
```

## Generar predicciones

**Nota:** Si modificáis algo del dataset de entreno, debéis aplicar la misma transformación al dataset de test. Esto es fundamental para que el modelo funcione correctamente.

```python
X_test = transform_data(test)
y_test = model.predict(X_test)

prediction = pd.DataFrame({"id": test["id"], "label": y_test})
prediction.to_csv("submission.csv", index=False)
```
